```python
import seaborn as sns
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
import pandas as pd
```

```python
iris = load_iris()
```

```python
X = iris.data
y = iris.target
```

```python
X.shape
```

```python
# here we have 4 feature we already knew it but in kmeans we can seperate data only on 2 featuer
```

```python
#visualize
sns.scatterplot(x = X[:,0], y = X[:,2])         #X[:,2] we select 2 feature which is 0 $ 2
```

# scaling

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# optinoal -- dimensional reduction using pca
```

```python
#elbow method
wcss = []
for k in range(1,11):
    kmeans = KMeans(n_clusters =k)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)
```

```python
sns.lineplot(x = range(1,11), y = wcss, marker = 'o')
```

```python
# kmeans
kmeans = KMeans(n_clusters = 3)
labels = kmeans.fit_predict(X_scaled)


sns.scatterplot(x= X[:,0] , y = X[:,2], c=labels)         # this is predict data we have
```

```python
sns.scatterplot(x= X[:,0] , y = X[:,2], c=y) #this is orginal data 
```

# by usng PCA 

```python
# by usng PCA 
from sklearn.decomposition import PCA

pca = PCA(n_components = 2)
pca_data = pca.fit_transform(X_scaled)
```

```python
#elbow method
wcss = []
for k in range(1,11):
    kmeans = KMeans(n_clusters =k)
    kmeans.fit(pca_data)
    wcss.append(kmeans.inertia_)
```

```python
sns.lineplot(x = range(1,11), y = wcss, marker = 'o')
```

```python
# kmeans
kmeans = KMeans(n_clusters = 3,random_state =10)
labels = kmeans.fit_predict(pca_data)


sns.scatterplot(x= pca_data[:,0] , y = pca_data[:,1], c=labels)# this is predict data we have
sns.scatterplot(x= kmeans.cluster_centers_[:,0] , y = kmeans.cluster_centers_[:,1], marker ='X' ,c="red", s=200)
```

```python
sns.scatterplot(x= X[:,0] , y = X[:,1], c=y)         # orignal data
```
